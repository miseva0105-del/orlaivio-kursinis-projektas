## 1. Pirmasis kodo blokas

**Įrankis:** ChatGPT → Colab.

**Įvestis:** 120 km nuotolis ir 130 km/h pastovus greitis.

**Išvestis:** 55,38 min trukmė.

Šis trumpas bandymas skirtas susipažinti su kodo vykdymu. Jis dar neaprašo aukštėjimo ir žemėjimo.

### Metodika: kelias, greitis ir trukmė

Kai greitis pastovus, skrydžio trukmė lygi nuotoliui, padalytam iš greičio. Šiame įvadiniame bandyme aukštėjimo ir žemėjimo etapai dar neatskiriami.

$$
t_{\mathrm{h}}=\frac{R}{V},\qquad \tau=60\frac{R}{V}
$$

R – horizontalusis nuotolis kilometrais; V – pastovus greitis km/h; tₕ – trukmė valandomis; τ – trukmė minutėmis. Skaitinis daugiklis 60 perskaičiuoja valandas į minutes.

$$
\tau=60\frac{120}{130}=55{,}38\ \mathrm{min}
$$



### Užklausa „ChatGPT“ kodui sukurti

```text
Sukurk Python kodą Google Colab užrašinei. Naudok lietuviškus komentarus ir antraštes. Pateik pilną kodo bloką. Duomenų neišgalvok ir nepriskirk mokomųjų prielaidų ICON A5 charakteristikoms. Apskaičiuok trukmę minutėmis, kai R_km=120 ir V_kmh=130: t_min=R_km/V_kmh*60. Išvesk rezultatą dviem skaitmenimis po kablelio.
```

Nukopijuokite užklausą į „ChatGPT“. Gautą kodą įklijuokite į „Colab“ kodo bloką ir paleiskite. Žemiau pateiktas patikrintas pavyzdinis kodas, kurį galima naudoti rezultatui palyginti.

````python
R_km = 120.0
V_kmh = 130.0
t_min = R_km / V_kmh * 60
print(f"Skrydžio trukmė: {t_min:.2f} min")
````

## 7. Skrydžio atkarpų skaičiavimas

**Įrankis:** ChatGPT → Colab.

**Įvestis:** Ankstesnio bloko p žodynas.

**Išvestis:** Trukmių ir horizontaliųjų nuotolių lentelė.

### Metodika: skrydžio atkarpų trukmės ir nuotoliai

Priimamas nulinis vėjas ir pastovus kiekvienos atkarpos tikrasis oro greitis (TAS). Visi aukščiai skaičiuojami nuo tos pačios atskaitos – jūros lygio. Aukštėjimo ir žemėjimo greičių dydžiai įvedami teigiamais skaičiais; kryptį nustato aukščio pokytis.

| Žymuo | Fizinė prasmė | Vienetas / ryšys su kodu |
| --- | --- | --- |
| V | Tikrasis oro greitis (TAS), naudojamas pradinėje lentelėje | km/h; V_climb_kmh, V_cruise_kmh, V_descent_kmh |
| v | Tas pats greitis, perskaičiuotas į SI vienetus | m/s |
| w | Aukštėjimo arba žemėjimo greičio dydis | m/s; w_climb_ms arba w_descent_ms |
| h₀, h₁ | Atkarpos pradinis ir galutinis aukščiai | m virš jūros lygio |
| t, τ | Tos pačios atkarpos trukmė sekundėmis ir minutėmis | t – s; τ – min |
| vₓ | Horizontalioji greičio dedamoji, kai vėjo nėra | m/s |
| R | Horizontalusis atkarpos nuotolis | km; bendras skrydžio nuotolis R_km |

### 1. Vienetų perskaičiavimas ir aukščio pokytis

$$
v=\frac{V}{3{,}6},\qquad \Delta h=h_1-h_0
$$

V formulėje įrašomas km/h, o v gaunamas m/s. Aukštėjant Δh > 0, žemėjant Δh < 0.

### 2. Aukštėjimo arba žemėjimo trukmė

$$
t=\frac{|\Delta h|}{w},\qquad \tau=\frac{t}{60}
$$

t gaunamas sekundėmis. Dalijant iš 60 gaunama tos pačios atkarpos trukmė minutėmis.

### 3. Horizontalioji greičio dedamoji ir nuotolis

$$
v^2=v_x^2+w^2,\qquad v_x=\sqrt{v^2-w^2}
$$

$$
R=\frac{v_x t}{1000}
$$

Nuotolis vₓt pirmiausia gaunamas metrais; dalijant iš 1000 jis perskaičiuojamas į kilometrus. Būtina sąlyga: 0 < w < v. Šis ryšys taikomas aukštėjimui ir žemėjimui, o kreiseriniame skrydyje w = 0.

### 4. Kreiserinio skrydžio nuotolis ir trukmė

$$
R_{\mathrm{kr}}=R_{\mathrm{bendr}}-R_{\mathrm{auk\check{s}t}}-R_{\mathrm{\check{z}em}}
$$

$$
t_{\mathrm{kr}}=3600\frac{R_{\mathrm{kr}}}{V_{\mathrm{kr}}},\qquad \tau_{\mathrm{kr}}=60\frac{R_{\mathrm{kr}}}{V_{\mathrm{kr}}}
$$

Rkr ir Rbendr pateikiami kilometrais, Vkr – km/h. Daugiklis 3600 paverčia valandas sekundėmis, o 60 – minutėmis. Rkr turi būti neneigiamas: jei aukštėjimas ir žemėjimas sunaudoja daugiau nuotolio nei numatyta, duomenų derinys netinka.

### 5. Rankinė aukštėjimo patikra

$$
t_{\mathrm{auk\check{s}t}}=\frac{600-0}{2}=300\ \mathrm{s}=5\ \mathrm{min}
$$

$$
R_{\mathrm{auk\check{s}t}}=\frac{\sqrt{(100/3{,}6)^2-2^2}\cdot300}{1000}\approx8{,}312\ \mathrm{km}
$$

Kontroliniai rezultatai: aukštėjimas – 8,312 km ir 5,000 min; kreiserinis skrydis – 99,481 km ir 45,914 min; žemėjimas – 12,207 km ir 6,667 min. Trijų skrydžio atkarpų suma – 120,000 km ir 57,581 min. Tai kinematinis įvertis; galios pakankamumas ir režimų pasiekiamumas dar netikrinami.



### Užklausa „ChatGPT“ kodui sukurti

```text
Sukurk Python kodą Google Colab užrašinei. Naudok lietuviškus komentarus ir antraštes. Pateik pilną kodo bloką. Duomenų neišgalvok ir nepriskirk mokomųjų prielaidų ICON A5 charakteristikoms. Ankstesniame bloke yra p žodynas. Parašyk funkciją skaiciuoti(p), naudojančią R_km, h_start_m, h_cruise_m, h_end_m, V_climb_kmh, w_climb_ms, V_cruise_kmh, V_descent_kmh, w_descent_ms. Laikyk TAS pastoviu atkarpoje, vėją nuliniu. t=abs(delta_h)/w; Vx=sqrt((TAS/3.6)**2-w**2); R=Vx*t/1000. Kreiseriniam skrydžiui skirk likusį R_km. Tikrink baigtinumą, teigiamus greičius, aukščių tvarką, w<TAS ir ar pakanka nuotolio. Grąžink pandas lentelę rez su stulpeliais: Etapas, Pradinis h, m; Galutinis h, m; TAS, km/h; Vertikalusis greitis, m/s; Trukmė, min; Nuotolis, km. Žemėjimo vertikaliąją dedamąją rodyk su minusu. Pridėk sukauptas trukmes ir nuotolius. Išvesk lentelę ir bendras sumas. Funkciją išlaikyk vėlesniam jautrumo bandymui.
```

Nukopijuokite užklausą į „ChatGPT“. Gautą kodą įklijuokite į „Colab“ kodo bloką ir paleiskite. Žemiau pateiktas patikrintas pavyzdinis kodas, kurį galima naudoti rezultatui palyginti.

````python
def skaiciuoti(p):
    keys = ['R_km', 'h_start_m', 'h_cruise_m', 'h_end_m',
            'V_climb_kmh', 'w_climb_ms', 'V_cruise_kmh',
            'V_descent_kmh', 'w_descent_ms']
    if any(not math.isfinite(float(p[k])) for k in keys):
        raise ValueError('Pradiniai duomenys turi būti baigtiniai skaičiai.')
    if p['R_km'] <= 0 or min(p['V_climb_kmh'], p['V_cruise_kmh'],
                            p['V_descent_kmh'], p['w_climb_ms'],
                            p['w_descent_ms']) <= 0:
        raise ValueError('Nuotolis ir greičiai turi būti teigiami.')
    if p['h_cruise_m'] <= max(p['h_start_m'], p['h_end_m']):
        raise ValueError('Kreiserinis aukštis turi būti didesnis už abiejų paviršių aukštį.')
    vc, vd = p['V_climb_kmh']/3.6, p['V_descent_kmh']/3.6
    wc, wd = p['w_climb_ms'], p['w_descent_ms']
    if wc >= vc or wd >= vd:
        raise ValueError('Vertikalusis greitis turi būti mažesnis už TAS.')
    tc = (p['h_cruise_m']-p['h_start_m'])/wc
    td = (p['h_cruise_m']-p['h_end_m'])/wd
    rc = math.sqrt(vc**2-wc**2)*tc/1000
    rd = math.sqrt(vd**2-wd**2)*td/1000
    rcr = p['R_km']-rc-rd
    if rcr < 0:
        raise ValueError(f'Aukštėjimui ir žemėjimui reikia {rc+rd:.2f} km; '
                         f'pasirinkta tik {p["R_km"]:.2f} km.')
    tcr = rcr/p['V_cruise_kmh']*3600
    rows = [
        ['Aukštėjimas', p['h_start_m'], p['h_cruise_m'], p['V_climb_kmh'], wc, tc/60, rc],
        ['Kreiserinis skrydis', p['h_cruise_m'], p['h_cruise_m'], p['V_cruise_kmh'], 0, tcr/60, rcr],
        ['Žemėjimas', p['h_cruise_m'], p['h_end_m'], p['V_descent_kmh'], -wd, td/60, rd]]
    return pd.DataFrame(rows, columns=['Etapas', 'Pradinis h, m', 'Galutinis h, m',
        'TAS, km/h', 'Vertikalusis greitis, m/s', 'Trukmė, min', 'Nuotolis, km'])

rez = skaiciuoti(p)
rez['Sukaupta trukmė, min'] = rez['Trukmė, min'].cumsum()
rez['Sukauptas nuotolis, km'] = rez['Nuotolis, km'].cumsum()
rodyti(rez.round(3))
print(f"Trijų modeliuojamų segmentų trukmė: {rez['Trukmė, min'].sum():.2f} min")
print(f"Horizontalusis nuotolis: {rez['Nuotolis, km'].sum():.2f} km")
````

### Metodika: judėjimas sausumoje ir vandenyje

Skaičiuojami horizontalieji judėjimo atstumai. Formulėse v₀ ir v₁ yra pradinio ir galutinio judėjimo greičio reikšmės m/s; įvesties greičius km/h prieš skaičiavimą reikia padalyti iš 3,6. t – sekundės, s – metrai.

### 1. Riedėjimas pastoviu greičiu

$$
s_{\mathrm{ried}}=v_{\mathrm{ried}}t_{\mathrm{ried}},\qquad R_{\mathrm{ried}}=\frac{s_{\mathrm{ried}}}{1000}
$$

$$
s_{\mathrm{ried}}=\frac{18}{3{,}6}\cdot60=300\ \mathrm{m}
$$

### 2. Įsibėgėjimas, kai pagreitis pastovus

$$
t_{\mathrm{greit}}=\frac{v_1-v_0}{a},\qquad s_{\mathrm{greit}}=\frac{v_1^2-v_0^2}{2a}
$$

a > 0 – pasirinktas pastovus pagreitis m/s²; v₁ > v₀. Pagal mokomąsias prielaidas v₀ atitinka 18 km/h, v₁ – 100 km/h, o a = 1,5 m/s².

$$
t_{\mathrm{greit}}=\frac{100/3{,}6-18/3{,}6}{1{,}5}\approx15{,}185\ \mathrm{s}
$$

$$
s_{\mathrm{greit}}=\frac{(100/3{,}6)^2-(18/3{,}6)^2}{2\cdot1{,}5}\approx248{,}868\ \mathrm{m}
$$

Gautą kinematinį kelią galima palyginti su 600 m turimu taku. Šis palyginimas neapima atitrūkimo dinamikos, kliūties perskridimo ar reikalaujamų atsargų, todėl nėra pilna kilimo charakteristikų patikra.

### 3. Lėtėjimas vandenyje

$$
b=|a_{\mathrm{l\dot{e}t}}|>0,\qquad t_{\mathrm{l\dot{e}t}}=\frac{v_0-v_1}{b},\qquad s_{\mathrm{l\dot{e}t}}=\frac{v_0^2-v_1^2}{2b}
$$

b – teigiamas lėtėjimo pagreičio dydis m/s². Greitis mažėja, todėl v₀ > v₁. Priėmus 110 → 18 km/h ir b = 1,0 m/s², gaunama t ≈ 25,556 s ir s ≈ 454,321 m. Tai mokomasis kinematinis modelis, ne gamintojo tūpimo charakteristika.

### 4. Aukštėjimo padalijimas į M-03 ir M-04

$$
t_{\mathrm{pr}}=\frac{h_{\mathrm{pr}}-h_0}{w_{\mathrm{auk\check{s}t}}},\qquad t_{\mathrm{lik}}=t_{\mathrm{auk\check{s}t}}-t_{\mathrm{pr}}
$$

hpr = 60 m – pasirinkta pradinio aukštėjimo pabaigos riba. Nuo 0 iki 60 m, kai aukštėjimo greitis 2 m/s, gaunama 30 s. Likusi aukštėjimo dalis priskiriama M-04. Ši riba nenustato važiuoklės įtraukimo komandos; ją lemia taikoma procedūra.

### Metodika: degalų tūris pagal atkarpos trukmę

Kiekvienai skrydžio atkarpai parenkamos pastovios valandinės degalų sąnaudos. Kai qᵢ pateikiamas litrais per valandą, o τᵢ – minutėmis, pirmiausia trukmė perskaičiuojama į valandas.

$$
Q_i=q_i\frac{\tau_i}{60},\qquad Q_{\mathrm{skryd}}=\sum_{i=1}^{3}Q_i
$$

| Žymuo | Reikšmė | Vienetas |
| --- | --- | --- |
| qᵢ | i-osios atkarpos valandinės sąnaudos | l/h |
| τᵢ | i-osios atkarpos trukmė | min |
| Qᵢ | i-ajai atkarpai sunaudojamas degalų tūris | l |
| Qskryd | Trijų skrydžio atkarpų degalų tūrių suma | l |

### Skaitinis kreiserinio skrydžio pavyzdys

$$
Q_{\mathrm{kr}}=18\frac{45{,}914}{60}\approx13{,}774\ \mathrm{l}
$$

Priėmus 22 l/h aukštėjant, 18 l/h kreiseriniame skrydyje ir 10 l/h žemėjant, visoms trims atkarpoms gaunama apie 16,719 l. Kodas naudoja nesuapvalintas trukmes.

$$
Q_{\mathrm{pap}}=18\frac{30}{60}=9\ \mathrm{l}

Šiam skaičiavimui reikia atkarpų trukmių ir valandinių sąnaudų. Greičio kitimo grafikas nereikalingas. Skrydžio greičiai jau buvo naudojami ankstesniame žingsnyje nuotoliams ir trukmėms nustatyti.

Naudok rez ir p q_climb_lh, q_cruise_lh, q_descent_lh, extra_min. Apskaičiuok trijų skrydžio atkarpų degalų tūrius pagal q*t_min/60; išvesk sanaudos lentelę ir sumą. Atskirai pateik papildomų minučių scenarijų. Tikrink reikšmių baigtinumą ir neneigiamumą. Rezervo reikalavimo neišgalvok.

Nukopijuokite užklausą į „ChatGPT“. Gautą kodą įklijuokite į „Colab“ kodo bloką ir paleiskite. Žemiau pateiktas patikrintas pavyzdinis kodas, kurį galima naudoti rezultatui palyginti.

````python
q = [p['q_climb_lh'], p['q_cruise_lh'], p['q_descent_lh']]
if any(v < 0 or not math.isfinite(v) for v in q) or p['extra_min'] < 0:
    raise ValueError('Sąnaudos ir papildoma trukmė turi būti neneigiami dydžiai.')
sanaudos = rez[['Etapas', 'Trukmė, min']].copy()
sanaudos['Sąnaudos, l/h'] = q
sanaudos['Degalų tūris, l'] = sanaudos['Trukmė, min']/60*sanaudos['Sąnaudos, l/h']
rodyti(sanaudos.round(3))
Q = sanaudos['Degalų tūris, l'].sum()
Q_extra = p['q_cruise_lh']*p['extra_min']/60
print(f'Trims segmentams: {Q:.2f} l')
print(f"Papildomam {p['extra_min']:.0f} min scenarijui: {Q_extra:.2f} l")
print(f'Tik šių dalių suma: {Q+Q_extra:.2f} l; tai nėra visos misijos degalų poreikis.')
````
