## 7. Calculate the flight segments

<p>Under zero wind, convert TAS from km/h to m/s. For climb and descent, divide the absolute altitude change by the vertical speed magnitude to get the duration in seconds. Calculate the horizontal speed by resolving TAS into horizontal and vertical components. Multiply by duration and convert the distance to kilometres. Use the remaining distance for level cruise. Check that the vertical speed is below TAS and that the distance available for cruise is non-negative.</p><p><strong>Check.</strong> The climb takes 5.000 min and covers 8.312 km, cruise takes 45.914 min and covers 99.481 km, and descent takes 6.667 min and covers 12.207 km. Total: 120.000 km in 57.581 min. Check the climb manually using 600 / 2 = 300 s.</p>

### Prompt for ChatGPT

```text
Write a complete Python cell for Google Colab with English comments and headings. Use the supplied values and identify teaching assumptions separately from documented ICON A5 specifications. Define skaiciuoti(p) using the existing input dictionary. For climb and descent use t = abs(delta_h) / w and R = sqrt((TAS / 3.6)**2 - w**2) * t / 1000. Allocate the remaining distance to cruise. Validate finite inputs, positive speeds, altitude ordering and enough horizontal distance. Return rez with Phase, Initial h, m, Final h, m, TAS, km/h, Vertical speed, m/s, Duration, min and Distance, km. Make the descent vertical speed negative, add cumulative totals and display the results.
```

````python
def skaiciuoti(p):
    keys = ['R_km', 'h_start_m', 'h_cruise_m', 'h_end_m',
            'V_climb_kmh', 'w_climb_ms', 'V_cruise_kmh',
            'V_descent_kmh', 'w_descent_ms']
    if any(not math.isfinite(float(p[k])) for k in keys):
        raise ValueError('Inputs must be finite numbers.')
    if p['R_km'] <= 0 or min(p['V_climb_kmh'], p['V_cruise_kmh'],
                            p['V_descent_kmh'], p['w_climb_ms'],
                            p['w_descent_ms']) <= 0:
        raise ValueError('Distance and speeds must be positive.')
    if p['h_cruise_m'] <= max(p['h_start_m'], p['h_end_m']):
        raise ValueError('Cruise altitude must be above both surface elevations.')
    vc, vd = p['V_climb_kmh']/3.6, p['V_descent_kmh']/3.6
    wc, wd = p['w_climb_ms'], p['w_descent_ms']
    if wc >= vc or wd >= vd:
        raise ValueError('Vertical speed must be below TAS.')
    tc = (p['h_cruise_m']-p['h_start_m'])/wc
    td = (p['h_cruise_m']-p['h_end_m'])/wd
    rc = math.sqrt(vc**2-wc**2)*tc/1000
    rd = math.sqrt(vd**2-wd**2)*td/1000
    rcr = p['R_km']-rc-rd
    if rcr < 0:
        raise ValueError(f'Climb and descent require {rc+rd:.2f} km; '
                         f'only {p["R_km"]:.2f} km.')
    tcr = rcr/p['V_cruise_kmh']*3600
    rows = [
        ['Climb', p['h_start_m'], p['h_cruise_m'], p['V_climb_kmh'], wc, tc/60, rc],
        ['Cruise', p['h_cruise_m'], p['h_cruise_m'], p['V_cruise_kmh'], 0, tcr/60, rcr],
        ['Descent', p['h_cruise_m'], p['h_end_m'], p['V_descent_kmh'], -wd, td/60, rd]]
    return pd.DataFrame(rows, columns=['Phase', 'Initial h, m', 'Final h, m',
        'TAS, km/h', 'Vertical speed, m/s', 'Duration, min', 'Distance, km'])

rez = skaiciuoti(p)
rez['Cumulative duration, min'] = rez['Duration, min'].cumsum()
rez['Cumulative distance, km'] = rez['Distance, km'].cumsum()
rodyti(rez.round(3))
print(f"Duration of the three modelled segments: {rez['Duration, min'].sum():.2f} min")
print(f"Horizontal distance: {rez['Distance, km'].sum():.2f} km")
````

<p>Extend the model to include taxiing at 18 km/h for 60 s, a take-off roll from 18 to 100 km/h at 1.5 m/s², initial climb to 60 m, continued climb and cruise, descent, and water deceleration from 110 to 18 km/h at 1.0 m/s².
````python
# Additional teaching assumptions for M-01, M-02, M-03 and M-06.
# These values are teaching assumptions for the kinematic model.
e = {
    'V_taxi_kmh': 18.0, 't_taxi_s': 60.0,
    'a_takeoff_ms2': 1.5,
    'h_initial_m': 60.0,
    'a_water_ms2': 1.0, 'V_water_end_kmh': 18.0
}
# To connect the model phases, the end speed of the take-off roll equals
# the climb TAS, and the initial speed on water equals the descent TAS.
# Zero wind. Actual lift-off and touchdown speeds require separate data.
for k, v in e.items():
    if not math.isfinite(v):
        raise ValueError(f'{k}: value must be finite.')
if min(e['V_taxi_kmh'], e['t_taxi_s'], e['a_takeoff_ms2'], e['a_water_ms2']) <= 0:
    raise ValueError('Taxi speed, time and acceleration magnitudes must be positive.')
if not p['h_start_m'] < e['h_initial_m'] < p['h_cruise_m']:
    raise ValueError('The initial climb boundary must lie between the initial and cruise altitudes.')
if not 0 <= e['V_water_end_kmh'] < p['V_descent_kmh']:
    raise ValueError('Final water speed must be non-negative and below the initial water speed.')
if not math.isfinite(p['runway_m']) or p['runway_m'] <= 0:
    raise ValueError('Runway length must be finite and positive.')
v0, vto = e['V_taxi_kmh']/3.6, p['V_climb_kmh']/3.6
if vto <= v0:
    raise ValueError('The end speed of the take-off roll must exceed taxi speed.')
t_roll = (vto-v0)/e['a_takeoff_ms2']
s_roll = (vto**2-v0**2)/(2*e['a_takeoff_ms2'])
t_initial = (e['h_initial_m']-p['h_start_m'])/p['w_climb_ms']
s_initial = math.sqrt(vto**2-p['w_climb_ms']**2)*t_initial/1000
vwater0, vwater1 = p['V_descent_kmh']/3.6, e['V_water_end_kmh']/3.6
t_water = (vwater0-vwater1)/e['a_water_ms2']
s_water = (vwater0**2-vwater1**2)/(2*e['a_water_ms2'])
misija6 = pd.DataFrame([
 ['M-01','Taxiing on land',e['t_taxi_s']/60,v0*e['t_taxi_s']/1000,'Extended'],
 ['M-02','Take-off roll and lift-off',t_roll/60,s_roll/1000,'Extended'],
 ['M-03','Initial climb',t_initial/60,s_initial,'Retract per POH; location unspecified'],
 ['M-04','Continued climb and cruise',
  rez.iloc[0]['Duration, min']-t_initial/60+rez.iloc[1]['Duration, min'],
  rez.iloc[0]['Distance, km']-s_initial+rez.iloc[1]['Distance, km'],'Retracted'],
 ['M-05','Descent and water approach',rez.iloc[2]['Duration, min'],rez.iloc[2]['Distance, km'],'Retracted'],
 ['M-06','Water landing and deceleration',t_water/60,s_water/1000,'Retracted']
], columns=['ID','Phase','Duration, min','Distance, km','Landing gear state'])
misija6['Cumulative duration, min'] = misija6['Duration, min'].cumsum()
misija6['Cumulative distance, km'] = misija6['Distance, km'].cumsum()
rodyti(misija6.round(3))
print(f'Take-off roll: {s_roll:.1f} m in {t_roll:.2f} s; available runway {p["runway_m"]:.0f} m.')
print('Kinematic roll within the available runway:', s_roll <= p['runway_m'])
print('Take-off safety and obstacle clearance require separate performance verification.')
print(f'Deceleration on water: {s_water:.1f} m in {t_water:.2f} s.')
print(f'Total horizontal distance for all modelled phases: {misija6["Distance, km"].sum():.3f} km; '
      f'duration: {misija6["Duration, min"].sum():.3f} min.')
e_units = {'V_taxi_kmh':'km/h','t_taxi_s':'s','a_takeoff_ms2':'m/s²',
           'h_initial_m':'m AMSL','a_water_ms2':'m/s²','V_water_end_kmh':'km/h'}
papildomi = pd.DataFrame([[k,v,e_units[k],'Teaching assumption'] for k,v in e.items()],
                         columns=['Code symbol','Value','Unit','Source type'])
````
